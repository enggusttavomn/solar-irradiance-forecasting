# Avaliação mensal corrigida e reproduzível

Este é o caderno de referência para o TCC e os artigos. Ele lê apenas os artefatos da avaliação mensal corrigida: treinamento de janeiro de 2020 a dezembro de 2023 (48 meses), teste retrospectivo walk-forward de janeiro a dezembro de 2024 (12 meses), alvo GHI contínuo e três baselines treinadas sem acesso ao teste. O min--max usa somente o trecho pré-teste disponível até dezembro de 2023, incluindo 2019 como contexto.

O protocolo usa o valor de referência NSRDB do mês anterior para prever o mês seguinte. Portanto, os números **não** representam uma previsão recursiva de 12 meses feita em dezembro de 2023. GHI é média de irradiância em W/m², incluindo horas noturnas; não é energia nem geração elétrica. Apenas os agregados diários e os metadados das consultas de 60 minutos foram preservados, não as respostas horárias brutas.


```python
from pathlib import Path
import pandas as pd

raiz = Path.cwd()
if not (raiz / 'resultados').exists():
    raiz = raiz.parent
base = raiz / 'resultados' / 'avaliacao_mensal_corrigida'
assert base.exists(), f'Execute treinar_todas_localidades.py antes: {base}'

resumo = pd.read_csv(base / 'resumo_modelos_mae.csv')
comparacao = pd.read_csv(base / 'comparacao_climatologia.csv')
localidades = pd.read_csv(base / 'resumo_localidades.csv')
sementes = pd.read_csv(base / 'variabilidade_sementes.csv')
protocolo = pd.read_csv(base / 'protocolo_temporal.csv')
```
